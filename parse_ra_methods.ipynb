```python
pattern = r'Table \d+\.\d+:'
table_num, table_description, code_type = None, None, None
data = []
for line in text.splitlines():
    if re.search(pattern, line) and 'SQL Programming' not in line:
        code_type = 'icd10-cm' if 'Diagnosis' in line else 'cpt'
        table_num, table_description = [x.strip() for x in line.split(':')]
    else:
        tokens = line.split()
        code = tokens[0].strip()
        description = ' '.join(tokens[1:]).strip()
        row = [code, description, code_type, table_num, table_description]
        data.append(row)
```

```python
df = spark.createDataFrame(data, 
                      ['code', 'description', 'code_type', 'table_num', 'table_description'])
df = (
    df.withColumn('mimi_src_file_date', F.lit(parse('2023-12-23').date()))
    .withColumn('mimi_src_file_name', F.lit('A-07-23-01213-Complete Report.pdf'))
    .withColumn('mimi_dlt_load_date', F.lit(datetime.today().date()))
)
```
